````python
import os
import uuid
from pathlib import Path
from dotenv import load_dotenv
from datetime import datetime

import pandas as pd
from PIL import Image

from langchain.tools import tool
from langchain_core.messages import HumanMessage
from langchain_core.documents import Document

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)

from langchain_chroma import Chroma

from langchain_text_splitters import RecursiveCharacterTextSplitter

print("✅ Librerías cargadas correctamente.")
````

````python
# ==========================================
# CONFIGURACIÓN DE GOOGLE GEMINI
# ==========================================

import os

GOOGLE_API_KEY = "TU_API_KEY_AQUI"

os.environ["GOOGLE_API_KEY"] = GOOGLE_API_KEY

print("✅ API Key cargada correctamente.")
````

````python
# ============================================================
# CREAR MODELO GEMINI Y EMBEDDINGS
# ============================================================

MODELO_LLM = "gemini-3.1-flash-lite"
MODELO_EMBEDDING = "gemini-embedding-2-preview"

llm = ChatGoogleGenerativeAI(
    model=MODELO_LLM,
    temperature=0
)

embeddings = GoogleGenerativeAIEmbeddings(
    model=MODELO_EMBEDDING
)

print("✅ Modelo Gemini y Embeddings creados correctamente.")
````

````python
# ============================================================
# CARGAR BASES DE CONOCIMIENTO
# ============================================================

from pathlib import Path

CONTRATOS_PATH = Path("01_Clausulas_Contractuales.txt")
PROTECCION_PATH = Path("02_Proteccion_Datos.txt")
CUMPLIMIENTO_PATH = Path("03_Cumplimiento_Etica.txt")

manual_contratos = CONTRATOS_PATH.read_text(encoding="utf-8")
politica_datos = PROTECCION_PATH.read_text(encoding="utf-8")
cumplimiento_etica = CUMPLIMIENTO_PATH.read_text(encoding="utf-8")

print("✅ Bases de conocimiento cargadas correctamente.\n")

print(f"📄 {CONTRATOS_PATH.name}")
print(f"Caracteres: {len(manual_contratos):,}")

print(f"\n📄 {PROTECCION_PATH.name}")
print(f"Caracteres: {len(politica_datos):,}")

print(f"\n📄 {CUMPLIMIENTO_PATH.name}")
print(f"Caracteres: {len(cumplimiento_etica):,}")
````

````python
# ============================================================
# DIVIDIR DOCUMENTOS EN CHUNKS
# ============================================================

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=400,
    chunk_overlap=50
)

chunks_contratos = text_splitter.create_documents([manual_contratos])
chunks_proteccion = text_splitter.create_documents([politica_datos])
chunks_cumplimiento = text_splitter.create_documents([cumplimiento_etica])

print("✅ Chunks creados correctamente.\n")

print(f"Contratos: {len(chunks_contratos)} chunks")
print(f"Protección de Datos: {len(chunks_proteccion)} chunks")
print(f"Cumplimiento: {len(chunks_cumplimiento)} chunks")

print("\nPrimer chunk de Contratos:\n")
print(chunks_contratos[0].page_content)
````

````python
# ============================================================
# CREAR BASES VECTORIALES (CHROMA)
# ============================================================

vectorstore_contratos = Chroma.from_documents(
    documents=chunks_contratos,
    embedding=embeddings,
    collection_name="contratos"
)

vectorstore_proteccion = Chroma.from_documents(
    documents=chunks_proteccion,
    embedding=embeddings,
    collection_name="proteccion_datos"
)

vectorstore_cumplimiento = Chroma.from_documents(
    documents=chunks_cumplimiento,
    embedding=embeddings,
    collection_name="cumplimiento"
)

print("✅ Bases vectoriales creadas correctamente.")
````

````python
# ============================================================
# CREAR RETRIEVERS
# ============================================================

retriever_contratos = vectorstore_contratos.as_retriever(
    search_kwargs={"k": 3}
)

retriever_proteccion = vectorstore_proteccion.as_retriever(
    search_kwargs={"k": 3}
)

retriever_cumplimiento = vectorstore_cumplimiento.as_retriever(
    search_kwargs={"k": 3}
)

print("✅ Retrievers creados correctamente.")
````

````python
# ============================================================
# AGENTE RAG - CONTRATOS
# ============================================================

def responder_contratos(pregunta):

    documentos = retriever_contratos.invoke(pregunta)

    contexto = "\n\n".join(
        doc.page_content
        for doc in documentos
    )

    prompt = f"""
Eres el Agente de Contratos de PATITO S.A.

Responde únicamente utilizando la información del contexto.

Si la respuesta no se encuentra en el contexto responde exactamente:

"No encontré esa información en la base de conocimiento de Contratos."

Contexto:

{contexto}

Pregunta:

{pregunta}
"""

    respuesta = llm.invoke(prompt)

    if isinstance(respuesta.content, list):
        texto = ""

        for bloque in respuesta.content:
            if isinstance(bloque, dict):
                texto += bloque.get("text", "")

        return texto

    return respuesta.content
````

````python
respuesta = responder_contratos(
    "¿Qué cláusulas mínimas debe contener un contrato de prestación de servicios?"
)

print(respuesta)
````

````python
# ============================================================
# FUNCIÓN GENÉRICA PARA AGENTES RAG
# ============================================================

def responder_rag(pregunta, retriever, nombre_agente):

    documentos = retriever.invoke(pregunta)

    contexto = "\n\n".join(
        doc.page_content
        for doc in documentos
    )

    prompt = f"""
Eres el Agente de {nombre_agente} de PATITO S.A.

Responde únicamente utilizando la información del contexto.

Si la respuesta no se encuentra en el contexto responde exactamente:

"No encontré esa información en la base de conocimiento de {nombre_agente}."

Contexto:

{contexto}

Pregunta:

{pregunta}
"""

    respuesta = llm.invoke(prompt)

    if isinstance(respuesta.content, list):
        texto = ""

        for bloque in respuesta.content:
            if isinstance(bloque, dict):
                texto += bloque.get("text", "")

        return texto

    return respuesta.content
````

````python
# ============================================================
# CREAR LOS TRES AGENTES RAG
# ============================================================

def agente_contratos(pregunta):
    return responder_rag(
        pregunta,
        retriever_contratos,
        "Contratos"
    )


def agente_proteccion(pregunta):
    return responder_rag(
        pregunta,
        retriever_proteccion,
        "Protección de Datos"
    )


def agente_cumplimiento(pregunta):
    return responder_rag(
        pregunta,
        retriever_cumplimiento,
        "Cumplimiento Normativo"
    )

print("✅ Los tres agentes fueron creados correctamente.")
````

````python
# ============================================================
# ARCHIVO DE REGISTRO DE SOLICITUDES LEGALES
# ============================================================

from pathlib import Path

REGISTRO_PATH = Path("registro_solicitudes_legal.txt")

if not REGISTRO_PATH.exists():
    REGISTRO_PATH.write_text(
        "",
        encoding="utf-8"
    )

print("✅ Archivo de registro listo.")
print(REGISTRO_PATH.resolve())
````

````python
# ============================================================
# AGENTE DE ACCIÓN
# REGISTRAR SOLICITUDES LEGALES
# ============================================================

from langchain.tools import tool
from datetime import datetime

contador = 1

@tool
def registrar_solicitud(
    tipo_contrato: str,
    proveedor: str,
    objeto: str,
    plazo: str,
    monto: str,
    trata_datos: str
):
    """
    Registra una solicitud legal en el archivo de solicitudes.
    """

    global contador

    solicitud_id = f"LEG-{contador:04d}"

    fecha = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    registro = (
        f"{solicitud_id} | "
        f"{fecha} | "
        f"{tipo_contrato} | "
        f"{proveedor} | "
        f"{objeto} | "
        f"{plazo} | "
        f"{monto} | "
        f"{trata_datos}\n"
    )

    with open(REGISTRO_PATH, "a", encoding="utf-8") as archivo:
        archivo.write(registro)

    contador += 1

    return f"✅ Solicitud registrada correctamente con ID {solicitud_id}"
````

````python
# ============================================================
# PROBAR EL AGENTE DE ACCIÓN
# ============================================================

respuesta = registrar_solicitud.invoke(
    {
        "tipo_contrato": "Prestación de Servicios",
        "proveedor": "ABC Consulting",
        "objeto": "Implementación ERP",
        "plazo": "12 meses",
        "monto": "USD 45000",
        "trata_datos": "Sí"
    }
)

print(respuesta)
````

````python
# ============================================================
# AGENTE MULTIMODAL
# ============================================================

import base64
from langchain_core.messages import HumanMessage


def agente_multimodal(ruta_imagen, pregunta):

    with open(ruta_imagen, "rb") as f:
        imagen_base64 = base64.b64encode(f.read()).decode("utf-8")

    mensaje = HumanMessage(
        content=[
            {
                "type": "text",
                "text": f"""
Eres un asistente legal especializado.

Analiza únicamente el documento mostrado en la imagen y responde:

{pregunta}

Si la información no aparece, indícalo claramente.
"""
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/png;base64,{imagen_base64}"
                }
            }
        ]
    )

    respuesta = llm.invoke([mensaje])

    if isinstance(respuesta.content, list):

        texto = ""

        for bloque in respuesta.content:

            if isinstance(bloque, dict):
                texto += bloque.get("text", "")

            else:
                texto += str(bloque)

        return texto

    return respuesta.content
````

````python
# ============================================================
# ORQUESTADOR - CLASIFICADOR DE INTENCIÓN
# ============================================================

def clasificar_intencion(pregunta):

    pregunta = pregunta.lower()

    # ---------- ACCIÓN (PRIMERO) ----------
    if any(p in pregunta for p in [
        "registrar",
        "registro",
        "crear solicitud",
        "nuevo contrato",
        "guardar contrato"
    ]):
        return "accion"

    # ---------- MULTIMODAL ----------
    elif any(p in pregunta for p in [
        "imagen",
        "foto",
        "escaneo",
        "escanear",
        "pdf",
        "documento"
    ]):
        return "multimodal"

    # ---------- CONTRATOS ----------
    elif any(p in pregunta for p in [
        "contrato",
        "cláusula",
        "clausula",
        "prestación",
        "servicio",
        "arrendamiento",
        "acuerdo"
    ]):
        return "contratos"

    # ---------- PROTECCIÓN ----------
    elif any(p in pregunta for p in [
        "dato",
        "datos",
        "protección",
        "proteccion",
        "privacidad",
        "rgpd",
        "consentimiento"
    ]):
        return "proteccion"

    # ---------- CUMPLIMIENTO ----------
    elif any(p in pregunta for p in [
        "ética",
        "etica",
        "cumplimiento",
        "compliance",
        "soborno",
        "corrupción",
        "conducta"
    ]):
        return "cumplimiento"

    return "general"
````

````python
# ============================================================
# ORQUESTADOR PRINCIPAL
# ============================================================

def orquestador(pregunta):

    agente = clasificar_intencion(pregunta)

    # ---------------- CONTRATOS ----------------
    if agente == "contratos":

        return agente_contratos(pregunta)

    # ---------------- PROTECCIÓN ----------------
    elif agente == "proteccion":

        return agente_proteccion(pregunta)

    # ---------------- CUMPLIMIENTO ----------------
    elif agente == "cumplimiento":

        return agente_cumplimiento(pregunta)

    # ---------------- ACCIÓN ----------------
    elif agente == "accion":

        return (
            "Has solicitado registrar un contrato.\n\n"
            "Utiliza la función registrar_solicitud() "
            "proporcionando los datos del contrato."
        )

    # ---------------- MULTIMODAL ----------------
    elif agente == "multimodal":

        return (
            "Has solicitado analizar una imagen.\n\n"
            "Utiliza la función:\n\n"
            'agente_multimodal("imagen.png", "tu pregunta")'
        )

    # ---------------- GENERAL ----------------
    else:

        return (
            "No pude identificar el tipo de consulta.\n"
            "Intenta realizar una pregunta relacionada con contratos, "
            "protección de datos, cumplimiento o análisis de imágenes."
        )
````

````python
print(
    orquestador(
        "¿Qué cláusulas debe contener un contrato de prestación de servicios?"
    )
)
````

````python
# ============================================================
# CHAT FINAL
# ============================================================

print("="*60)
print("      ASISTENTE LEGAL INTELIGENTE")
print("="*60)
print("Escriba 'salir' para finalizar.\n")

while True:

    pregunta = input("👤 Usuario: ")

    if pregunta.lower() == "salir":
        print("\n👋 Gracias por utilizar el Asistente Legal.")
        break

    tipo = clasificar_intencion(pregunta)

    # --------- AGENTE MULTIMODAL ---------
    if tipo == "multimodal":

        ruta = input("📷 Nombre de la imagen: ")
        consulta = input("❓ Consulta sobre la imagen: ")

        print("\n🤖 Asistente:\n")
        print(agente_multimodal(ruta, consulta))
        print("\n" + "-"*60 + "\n")
        continue

    # --------- AGENTE ACCIÓN ---------
    if tipo == "accion":

        print("\nRegistro de nueva solicitud legal\n")

        tipo_contrato = input("Tipo de contrato: ")
        proveedor = input("Proveedor: ")
        objeto = input("Objeto: ")
        plazo = input("Plazo: ")
        monto = input("Monto: ")
        trata_datos = input("¿Trata datos personales? (Sí/No): ")

        datos = {
            "tipo_contrato": tipo_contrato,
            "proveedor": proveedor,
            "objeto": objeto,
            "plazo": plazo,
            "monto": monto,
            "trata_datos": trata_datos
        }

        print("\n🤖 Asistente:\n")

        resultado = registrar_solicitud.invoke(datos)

        print(resultado)

        print("\n" + "-"*60 + "\n")
        continue

    # --------- AGENTES RAG ---------

    print("\n🤖 Asistente:\n")
    print(orquestador(pregunta))
    print("\n" + "-"*60 + "\n")
````

# Conclusiones del Proyecto

## Descripción del proyecto

Este proyecto consiste en el desarrollo de un **Asistente Legal Inteligente** basado en Inteligencia Artificial utilizando **LangChain**, **Google Gemini** y **ChromaDB**. La solución integra diferentes agentes especializados que permiten responder consultas legales mediante Recuperación Aumentada por Generación (RAG), analizar documentos en formato de imagen mediante capacidades multimodales y registrar solicitudes legales utilizando herramientas de acción. Todo el sistema es coordinado por un clasificador de intenciones y un orquestador que dirigen automáticamente cada consulta al agente más adecuado, proporcionando una experiencia de interacción unificada.

---

## Objetivos cumplidos

Durante el desarrollo del proyecto se alcanzaron los siguientes objetivos:

- ✅ Implementación de un sistema RAG especializado en **cumplimiento normativo y ética empresarial**.

- ✅ Implementación de un **Agente Multimodal** para analizar imágenes de documentos legales mediante el modelo Gemini.
- ✅ Desarrollo de un **clasificador de intenciones** que identifica automáticamente el tipo de consulta realizada por el usuario.
- ✅ Implementación de un **orquestador** que dirige cada solicitud hacia el agente especializado correspondiente.
